# 1- TF-IDF

<img src="https://drive.google.com/uc?export=view&id=1saSt0mMgOQ2ybbTms1lu_ruhUcBWkKzL" width="500">

TF-IDF stands for term frequency-inverse document frequency. It is a measure that discounts common words. Used in the fields of information retrieval (IR) and machine learning, that can quantify the importance of words in a document amongst a collection of documents (also known as a corpus).

### Components of TF-IDF

1. **TF (Term Frequency):**  
   Measures how frequently a term appears in a document.


$$ \text{TF}(t, d) = \frac{\text{Number of times term } t \text{ appears in document } d}{\text{Total number of terms in document } d} $$


2. **IDF (Inverse Document Frequency):**  
   Measures how important a term is across all documents. Words that appear in many documents get lower scores.

$$    \text{IDF}(t) = \log \left(\frac{\text{Number of all documents N}}{\text{Number of documents containing the term } t}\right) $$

# Cosine similarity

In NLP, Cosine similarity is a metric used to measure how similar the documents are.


$$ \text{cosine similarity} = \frac{A \cdot B}{\|A\| \times \|B\|} $$

Where:  
$ A \cdot B $  = dot product of vectors A and B  
$ \|A\| $ = magnitude (length) of vector A  
$ \|B\| $ =  magnitude (length) of vector B

#### Intuition

- If vectors point in the **same direction**, cosine similarity = **1** (maximum similarity).
- If vectors are **orthogonal (90° apart)**, cosine similarity = **0** (no similarity).

### Task 1: Cosine Similarity
Use the Cosine Similarity method to determine how similar the following sentences are.

'This is the first document.',  
'This document is the second document.',  
'And this is the third one.',  
'Is this the first document?'  


````python
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

task1_sentences = [
    "This is the first document.",
    "This document is the second document.",
    "And this is the third one.",
    "Is this the first document?"
]
task1_vectors = TfidfVectorizer().fit_transform(task1_sentences)
task1_similarity = cosine_similarity(task1_vectors)
task1_results = pd.DataFrame(
    task1_similarity,
    index=[f"Sentence {i}" for i in range(1, 5)],
    columns=[f"Sentence {i}" for i in range(1, 5)]
)
display(task1_results.round(4))
print("Most similar distinct sentences: Sentence 1 and Sentence 4 "
      f"(cosine similarity = {task1_results.iloc[0, 3]:.4f}).")
````

### Task 2: TF-IDF
Use tf-idf method on the sentences below to determine the important words.

'data science is one of the most important fields of science',  
'this is one of the best data science courses',  
'data scientists analyze data'  


````python
from sklearn.feature_extraction.text import TfidfVectorizer
import pandas as pd

task2_sentences = [
    "data science is one of the most important fields of science",
    "this is one of the best data science courses",
    "data scientists analyze data"
]
task2_tfidf = TfidfVectorizer(stop_words="english")
task2_vectors = task2_tfidf.fit_transform(task2_sentences)
task2_scores = pd.DataFrame(
    task2_vectors.toarray(),
    index=[f"Sentence {i}" for i in range(1, 4)],
    columns=task2_tfidf.get_feature_names_out()
)
display(task2_scores.round(4))

for sentence, scores in task2_scores.iterrows():
    important_words = scores[scores > 0].sort_values(ascending=False)
    print(f"{sentence} - important words (highest TF-IDF first):")
    print(important_words.round(4).to_string())
    print()
````

````python
import re
from pathlib import Path

import pandas as pd
import gensim
from nltk.tokenize import wordpunct_tokenize

# This is the public CSV path listed by the TidyTuesday Simpsons dataset project.
DATASET_URL = (
    "https://raw.githubusercontent.com/rfordatascience/tidytuesday/"
    "main/data/2025/2025-02-04/simpsons_script_lines.csv"
)


simpsons_path = next(
    (path for path in (Path("simpsons_script_lines.csv"),
                       Path("/content/simpsons_script_lines.csv"),
                       Path("/mnt/data/simpsons_script_lines.csv"))
     if path.is_file()),
    None
)
try:
    simpsons_df = pd.read_csv(
        simpsons_path if simpsons_path is not None else DATASET_URL,
        usecols=["spoken_words"],
        on_bad_lines="skip",
        low_memory=False
    )
except Exception as exc:
    raise FileNotFoundError(
        "Could not read simpsons_script_lines.csv. Download it from the "
        "Simpsons dataset source, place it next to this notebook (or upload "
        "it to Colab), and rerun this cell. An internet connection is "
        "needed only when downloading it automatically."
    ) from exc

# Use the cleaning procedure supplied in Task 3.
def clean_text(text):
    text = text.lower()
    text = re.sub(r"[0-9]", "", text)
    text = re.sub(r"[)(,”“.’$-]", "", text)
    return text

# Remove missing/empty lines, then tokenize each spoken line as its own document.
spoken_lines = simpsons_df["spoken_words"].dropna().astype(str)
spoken_lines = spoken_lines.map(clean_text)
tokens = [
    [word for word in wordpunct_tokenize(line) if word.isalpha()]
    for line in spoken_lines
]
tokens = [line_tokens for line_tokens in tokens if line_tokens]

# Use the EXACT Word2Vec configuration requested in the assignment.
Skip_gram_model = gensim.models.Word2Vec(
    tokens, min_count=1, vector_size=100, window=5, sg=1
)
print(f"Training complete: {len(tokens):,} dialogue lines; "
      f"{len(Skip_gram_model.wv):,} words in the vocabulary.")
````

### Task 4

Use: wv.most_similar() method to :

1.	Find the words similar to “homer”.

2.	Find the words similar to “marge”.

3. Find the words similar to “bart”



````python
for character in ["homer", "marge", "bart"]:
    if character in Skip_gram_model.wv:
        print(f"Words most similar to '{character}':")
        for word, score in Skip_gram_model.wv.most_similar(character):
            print(f"  {word:20s} {score:.4f}")
    else:
        print(f"'{character}' is not in the trained model's vocabulary.")
    print()
````
